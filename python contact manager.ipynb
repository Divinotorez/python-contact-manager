```python
contacts=list()                    #liste permettant de stocker les contacts
Nom=set()    #ensemble permettant de stocker les noms enregistrer
```

# **Ajouter un  contact**


```python
#création d'une fonction ajouter pour faciliter la mise à jour
def ajouter() :
    #démander à l'utilisateur ses information
    print("Entrez les informations :\n")
    nom=str(input("Nom:"))
    for noms in Nom :
        if nom!=noms :
          tel=(input("Téléphone:"))
          mail=str(input("Email:"))
          #stcoker les information de chacun dans un dictionnaire
          mon_dict={"Nom": nom,"Téléphone": tel,"Email":mail}
          contacts.append(mon_dict)             # Ajouter le contact créer à la liste 
          Nom.add(nom)     
        else:    
           print("Se contact est déjà enregistrer dans le carnat :\n")
           recherche(nom) 
ajouter()  
```

# **Afficher les contacts**

```python
#affichage des contacts
print(contacts)
```

# **Rechercher un contact à partir de son nom**

```python
#on crèer une fonction pour faciliter l'automatisation
def recherche(nom):
    for i in range(len(contacts)):
            if contacts[i]["Nom"]==nom:
                print("Voici les informations sur le contact")
                print(contacts[i])
            else :
                     print("Désolé! Cette personne ne se trouve pas dans notre carnet")
 
```

```python
#Effectuer une recherche
nom_chercher=input("Enter le nom du contact que vous rechercher")
recherche(nom_chercher)
```

# **Supprimer un contact à partir de son nom**

```python
#On crèe un fonction supprimer pour faciliter l'automatisation
def suppression():
    nom=str(input("Entrer le nom du contact que vous souhaitez supprimer "))
    for i in range(len(contacts)) :
            if contacts[i]["Nom"]==nom:
                del contacts[i]
                f"{nom} a été supprimé du carnet"
            else :
                     print("Désolé! Cette personne ne se trouve pas dans notre carnet")
suppression()  #Exécution de la suppression
```

### **Sauvegarder les contacts ligne par ligne dans le fichier contact**

```python
def sauvegarde():
  with open ("contact.txt","a") as f:
     for i in range (len(contacts)) :
        contact= contacts[i]["Nom"] + "," + contacts[i]["Téléphone"] + "," + contacts[i]["Email"]
        f.write(contact + "\n")
        
                   
```
